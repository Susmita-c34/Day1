# f-**string** **bold text**
The f before the opening quote makes it an f-string. Anything inside { } is replaced by its value.

```python
name='susm'
age=22
cgpa= 3.5666
print(f"Her name was {name} and I know her age is {age}")
print(f"CGPA{cgpa:.3f}")
print(f"next year her age will be{age+1}")
```

# Student score card

```python
name=input("name of the student?  ")
#inputs of marks
math_sc=float(input("Math marks:  "))
science_sc=float(input("Science marks:  "))
english_sc=float(input("English marks:  "))
#calculations
total= math_sc+science_sc+english_sc
average= total/3
highest= max(math_sc,science_sc,english_sc)
#output
boarder= "=" * 32
print(border)
print("SCORE CARD")
print(border)
print(f"Student: {name}")
print(f"Math   :  {math_sc:.2f}")
print(f"Science: {science_sc:.2f}")
print(f"English: {english_sc:.2f}")
print("-" *32)
print(f"Total  : {total:.2f}")
print(f"Average: {average:.2f}")
print(f"Highest: {highest:.2f}")
print(border)
```

```python
sent= " I love machine learning "
cleaned_sent= sent.strip().lower()
words = sent.split()
char_count =len(cleaned_sent.replace(" ",""))
word_count =len(words)
average_word_length = char_count / word_count
rev_sent= cleaned_sent[::-1]
border = "="*20
print(border +"TEXT ANALYSIS"+border)
print("Original       :",sent)
print("Cleaned        :",cleaned_sent)
print("Word count     :",word_count)
print("Char count     :",char_count)
print(f"Avg wordlen    :{average_word_length:.2f}")
print("Reversed       :",rev_sent)
print("Contain 'machine'?","machine" in sent)
print(border+border+border)
```
